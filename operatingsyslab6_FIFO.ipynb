```python
def fifo_page_replacement():
    pages=list(map(int,input("Enter reference string(seperated by space): ").split()))
    capacity = int(input("Enter number of frames: "))
    memory = []
    page_faults = 0

    print("\n--- Execution Steps ---")
    for page in pages:
        if page not in memory:
            if len(memory) == capacity:
                memory.pop(0)
            memory.append(page)
            page_faults+=1
            print(f"Page {page} -> Fault | Frames: {memory}")
        else:
            print(f"Page {page} -> Hit | Frames: {memory}")

    print(f"\nTotal Page Faults: {page_faults}")
    print(f"Toatl Page Hits: {len(pages)-page_faults}")

fifo_page_replacement()
```

```python
def mru_paging():
    capacity = int(input("Enter number of frames: "))
    pages = input("Enter page reference string (space-separated): ").split()

    frames = []
    page_faults = 0
    page_hits=0

    print("\nStep-by-step execution:")

    for page in pages:
        if page not in frames:
            page_faults += 1
            status = "Miss (Fault)"

            if len(frames) < capacity:
                frames.append(page)
            else:
                frames.pop()
                frames.append(page)
        else:
            page_hits += 1
            status="Hit"
            frames.remove(page)
            frames.append(page)

        print(f"Page: {page:2} | Frames: {str(frames):<15} | {status}")

    print(f"\nTotal Page Faults: {page_faults}")
    print(f"Toatl Page Hits: {page_hits}")

if __name__ =="__main__":
  mru_paging()
```
